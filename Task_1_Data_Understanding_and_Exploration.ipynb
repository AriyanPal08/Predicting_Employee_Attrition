# Task 2 - Data Preprocessing

This section prepares the Task 1 dataset for later model development. No model is trained here. All transformations are fitted using the training set only, which prevents information from the test set leaking into preprocessing.

### Preprocessing decisions

- **Missing values:** Task 1 found no missing values, so no imputation is required. This is verified again below.
- **Dropped columns:** `EmployeeCount`, `Over18`, and `StandardHours` are constant for every employee and cannot distinguish between attrition outcomes. `EmployeeNumber` is a unique employee identifier rather than an employee characteristic, so it is also excluded.
- **Target encoding:** `Attrition` is converted from `No`/`Yes` to `0`/`1`; the original target labels are not altered in `df`.
- **Categorical encoding:** One-hot encoding is used for nominal categories because their labels have no numeric order. `handle_unknown='ignore'` safely handles a category that may appear later but was not present during training.
- **Feature scaling:** Standard scaling is applied only to numerical predictors. It puts variables measured on different scales onto a comparable scale for scale-sensitive models such as Logistic Regression. One-hot encoded columns are not scaled.
- **Train-test split:** An 80/20 stratified split is used with `random_state=42`. Stratification retains the minority attrition proportion in both sets.

```python
# Import only the preprocessing tools needed for Task 2.
from sklearn.compose import ColumnTransformer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder, StandardScaler

RANDOM_STATE = 42
TEST_SIZE = 0.20

print('Task 2 configuration:')
print(f'Random state: {RANDOM_STATE}')
print(f'Test-set proportion: {TEST_SIZE:.0%}')
print(f'Training-set proportion: {1 - TEST_SIZE:.0%}')
```

```python
# 1. Reconfirm missing values before preprocessing.
print('=' * 70)
print('TASK 2.1 - MISSING-VALUE VERIFICATION')
print('=' * 70)

missing_value_counts = df.isna().sum()
print(f'Total missing values in the dataset: {missing_value_counts.sum()}')
print('Missing-value counts by column:')
print(missing_value_counts.to_string())

if missing_value_counts.sum() == 0:
    print('Decision: No imputation is performed because the dataset has no missing values.')
else:
    raise ValueError('Missing values were found. Review the data before continuing.')
```

```python
# 2. Remove columns that cannot provide useful predictive information.
print('=' * 70)
print('TASK 2.2 - DROP CONSTANT AND IDENTIFIER COLUMNS')
print('=' * 70)

constant_columns_to_drop = ['EmployeeCount', 'Over18', 'StandardHours']
identifier_columns_to_drop = ['EmployeeNumber']
columns_to_drop = constant_columns_to_drop + identifier_columns_to_drop

missing_expected_columns = [
    column for column in columns_to_drop if column not in df.columns
]
if missing_expected_columns:
    raise ValueError(f'Expected columns were not found: {missing_expected_columns}')

print('Constant columns removed because they have one value for every employee:')
for column in constant_columns_to_drop:
    print(f'- {column}: unique values = {df[column].nunique(dropna=False)}')

print('Identifier columns removed because they identify rows rather than employee attributes:')
for column in identifier_columns_to_drop:
    print(f'- {column}: unique values = {df[column].nunique(dropna=False)}')

X = df.drop(columns=[TARGET_COLUMN] + columns_to_drop).copy()
print(f'Feature matrix shape after dropping columns: {X.shape}')
print(f'Remaining predictor columns ({X.shape[1]}):')
print(X.columns.tolist())
```

```python
# 3. Encode the binary target without modifying the original df column.
print('=' * 70)
print('TASK 2.3 - TARGET ENCODING')
print('=' * 70)

target_mapping = {'No': 0, 'Yes': 1}
observed_target_values = set(df[TARGET_COLUMN].dropna().unique())
unexpected_target_values = observed_target_values - set(target_mapping)

if unexpected_target_values:
    raise ValueError(f'Unexpected target values found: {unexpected_target_values}')

y = df[TARGET_COLUMN].map(target_mapping)

print(f'Target mapping: {target_mapping}')
print(f'Target vector shape: {y.shape}')
print('Encoded target value counts:')
print(y.value_counts().sort_index().to_string())

if y.isna().any():
    raise ValueError('Target encoding produced missing values.')
```

```python
# 4. Split the raw data before fitting encoders or scalers.
# Stratification preserves the attrition class balance in both partitions.
print('=' * 70)
print('TASK 2.4 - STRATIFIED TRAIN-TEST SPLIT')
print('=' * 70)

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=TEST_SIZE,
    random_state=RANDOM_STATE,
    stratify=y
)

print(f'Raw X_train shape: {X_train.shape}')
print(f'Raw X_test shape: {X_test.shape}')
print(f'y_train shape: {y_train.shape}')
print(f'y_test shape: {y_test.shape}')

print('\ny_train class counts:')
print(y_train.value_counts().sort_index().to_string())
print('y_train class percentages:')
print((y_train.value_counts(normalize=True).sort_index() * 100).round(2).to_string())

print('\ny_test class counts:')
print(y_test.value_counts().sort_index().to_string())
print('y_test class percentages:')
print((y_test.value_counts(normalize=True).sort_index() * 100).round(2).to_string())
```

```python
# 5. Identify categorical and numerical features from the training predictors.
print('=' * 70)
print('TASK 2.5 - FEATURE GROUPS FOR TRANSFORMATION')
print('=' * 70)

categorical_features_for_encoding = X_train.select_dtypes(
    include=['object', 'string', 'category', 'bool']
).columns.tolist()

numerical_features_for_scaling = X_train.select_dtypes(
    include='number'
).columns.tolist()

print(f'Categorical features to one-hot encode ({len(categorical_features_for_encoding)}):')
print(categorical_features_for_encoding)
print(f'\nNumerical features to standard-scale ({len(numerical_features_for_scaling)}):')
print(numerical_features_for_scaling)

if not categorical_features_for_encoding or not numerical_features_for_scaling:
    raise ValueError('Both categorical and numerical feature groups are expected.')
```

```python
# 6. Encode categorical variables and scale numerical variables.
# fit_transform is used only on X_train to avoid data leakage.
print('=' * 70)
print('TASK 2.6 - ENCODING AND SCALING')
print('=' * 70)

preprocessor = ColumnTransformer(
    transformers=[
        (
            'categorical_encoder',
            OneHotEncoder(handle_unknown='ignore', sparse_output=False),
            categorical_features_for_encoding
        ),
        (
            'numerical_scaler',
            StandardScaler(),
            numerical_features_for_scaling
        )
    ],
    remainder='drop',
    verbose_feature_names_out=False
)

X_train_processed_array = preprocessor.fit_transform(X_train)
X_test_processed_array = preprocessor.transform(X_test)
processed_feature_names = preprocessor.get_feature_names_out()

X_train_processed = pd.DataFrame(
    X_train_processed_array,
    columns=processed_feature_names,
    index=X_train.index
)
X_test_processed = pd.DataFrame(
    X_test_processed_array,
    columns=processed_feature_names,
    index=X_test.index
)

print(f'Number of processed features: {len(processed_feature_names)}')
print(f'Processed X_train shape: {X_train_processed.shape}')
print(f'Processed X_test shape: {X_test_processed.shape}')
print(f'Missing values in processed X_train: {X_train_processed.isna().sum().sum()}')
print(f'Missing values in processed X_test: {X_test_processed.isna().sum().sum()}')
print('\nProcessed feature names:')
print(processed_feature_names.tolist())
```

```python
# 7. Verify scaling and final Task 2 outputs.
print('=' * 70)
print('TASK 2.7 - FINAL PREPROCESSING VERIFICATION')
print('=' * 70)

scaling_verification = pd.DataFrame({
    'Training Mean': X_train_processed[numerical_features_for_scaling].mean(),
    'Training Standard Deviation': X_train_processed[numerical_features_for_scaling].std()
}).round(4)

print('Scaled numerical-feature verification (training data only):')
print(scaling_verification.to_string())

print('\nFinal shapes after preprocessing:')
print(f'X_train_processed: {X_train_processed.shape}')
print(f'X_test_processed: {X_test_processed.shape}')
print(f'y_train: {y_train.shape}')
print(f'y_test: {y_test.shape}')

print('\nFinal class distribution in the training set:')
print(pd.DataFrame({
    'Count': y_train.value_counts().sort_index(),
    'Percentage (%)': (y_train.value_counts(normalize=True).sort_index() * 100).round(2)
}).to_string())

print('\nFinal class distribution in the test set:')
print(pd.DataFrame({
    'Count': y_test.value_counts().sort_index(),
    'Percentage (%)': (y_test.value_counts(normalize=True).sort_index() * 100).round(2)
}).to_string())

print('\nTask 2 complete. No model training has been performed.')
```

### Task 2 conclusion

The dataset is now split into stratified training and test sets. The fitted `preprocessor` and the processed feature matrices (`X_train_processed` and `X_test_processed`) are ready for Task 3 model development.

# Task 3 - Model Development

This section trains four classification models using the processed training data created in Task 2. It generates and stores test-set class predictions only. Model evaluation is intentionally deferred to Task 4.

### Model and hyperparameter choices

1. **Logistic Regression:** `solver='lbfgs'`, `max_iter=1000`, and `random_state=42`. The `lbfgs` solver is suitable for binary classification, while 1,000 iterations gives the optimiser sufficient time to converge on the processed features.
2. **Decision Tree:** `criterion='gini'`, `max_depth=5`, `min_samples_split=10`, `min_samples_leaf=5`, and `random_state=42`. The depth and minimum-sample limits constrain tree complexity and reduce the chance of learning overly specific training patterns.
3. **Random Forest:** `n_estimators=300`, `criterion='gini'`, `max_depth=10`, `min_samples_split=10`, `min_samples_leaf=4`, `class_weight='balanced'`, `random_state=42`, and `n_jobs=-1`. A larger collection of constrained trees provides a more stable ensemble, while balanced class weights account for the lower number of attrition cases in the training set.
4. **Bagging Classifier:** 100 decision-tree estimators, each with `max_depth=5`, `min_samples_split=10`, `min_samples_leaf=5`, and `random_state=42`; the bagging classifier uses `max_samples=1.0`, `max_features=1.0`, `random_state=42`, and `n_jobs=-1`. Bagging combines bootstrap-trained trees to reduce the instability of one decision tree.

```python
# Import only the four permitted model classes used in Task 3.
from sklearn.ensemble import BaggingClassifier, RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier

print('=' * 70)
print('TASK 3 - MODEL DEVELOPMENT SETUP')
print('=' * 70)

required_task2_objects = [
    'X_train_processed', 'X_test_processed', 'y_train', 'y_test', 'RANDOM_STATE'
]
missing_task2_objects = [
    object_name for object_name in required_task2_objects
    if object_name not in globals()
]

if missing_task2_objects:
    raise RuntimeError(
        'Run all Task 1 and Task 2 cells before Task 3. Missing objects: '
        f'{missing_task2_objects}'
    )

if RANDOM_STATE != 42:
    raise ValueError(f'Expected RANDOM_STATE to be 42, found {RANDOM_STATE}.')

print(f'X_train_processed shape: {X_train_processed.shape}')
print(f'X_test_processed shape: {X_test_processed.shape}')
print(f'y_train shape: {y_train.shape}')
print(f'y_test shape: {y_test.shape}')
print('\nTraining target class counts:')
print(y_train.value_counts().sort_index().to_string())
print('\nTest target class counts:')
print(y_test.value_counts().sort_index().to_string())
```

### Model 1 - Logistic Regression

Logistic Regression is a linear binary classifier. The numerical features were already standard-scaled in Task 2, which is appropriate because this model is sensitive to feature scale.

```python
# Model 1: Logistic Regression
logistic_regression_model = LogisticRegression(
    solver='lbfgs',
    max_iter=1000,
    random_state=RANDOM_STATE
)

print('Logistic Regression hyperparameters:')
print(logistic_regression_model.get_params())

logistic_regression_model.fit(X_train_processed, y_train)
logistic_regression_predictions = logistic_regression_model.predict(X_test_processed)

print('Logistic Regression has been fitted successfully.')
print('Logistic Regression predictions have been generated.')
print(f'Prediction shape: {logistic_regression_predictions.shape}')
print('Predicted class counts (not evaluation metrics):')
print(pd.Series(logistic_regression_predictions).value_counts().sort_index().to_string())
```

### Model 2 - Decision Tree

A Decision Tree learns rule-based splits. Its depth and minimum-sample settings are deliberately limited so that the individual tree remains easier to interpret and less likely to memorise small groups of employees.

```python
# Model 2: Decision Tree
decision_tree_model = DecisionTreeClassifier(
    criterion='gini',
    max_depth=5,
    min_samples_split=10,
    min_samples_leaf=5,
    random_state=RANDOM_STATE
)

print('Decision Tree hyperparameters:')
print(decision_tree_model.get_params())

decision_tree_model.fit(X_train_processed, y_train)
decision_tree_predictions = decision_tree_model.predict(X_test_processed)

print('Decision Tree has been fitted successfully.')
print('Decision Tree predictions have been generated.')
print(f'Prediction shape: {decision_tree_predictions.shape}')
print('Predicted class counts (not evaluation metrics):')
print(pd.Series(decision_tree_predictions).value_counts().sort_index().to_string())
```

### Model 3 - Random Forest

A Random Forest averages predictions from many decision trees. Class weights are balanced because attrition cases are the smaller class in the Task 2 training data.

```python
# Model 3: Random Forest
random_forest_model = RandomForestClassifier(
    n_estimators=300,
    criterion='gini',
    max_depth=10,
    min_samples_split=10,
    min_samples_leaf=4,
    class_weight='balanced',
    random_state=RANDOM_STATE,
    n_jobs=-1
)

print('Random Forest hyperparameters:')
print(random_forest_model.get_params())

random_forest_model.fit(X_train_processed, y_train)
random_forest_predictions = random_forest_model.predict(X_test_processed)

print('Random Forest has been fitted successfully.')
print('Random Forest predictions have been generated.')
print(f'Prediction shape: {random_forest_predictions.shape}')
print('Predicted class counts (not evaluation metrics):')
print(pd.Series(random_forest_predictions).value_counts().sort_index().to_string())
```

### Model 4 - Bagging Classifier

Bagging trains several decision trees on bootstrap samples and combines their predictions. The base trees use the same constraints as the standalone tree so that the ensemble does not rely on highly complex individual trees.

```python
# Model 4: Bagging Classifier with constrained decision trees.
bagging_base_estimator = DecisionTreeClassifier(
    criterion='gini',
    max_depth=5,
    min_samples_split=10,
    min_samples_leaf=5,
    random_state=RANDOM_STATE
)

bagging_classifier_model = BaggingClassifier(
    estimator=bagging_base_estimator,
    n_estimators=100,
    max_samples=1.0,
    max_features=1.0,
    random_state=RANDOM_STATE,
    n_jobs=-1
)

print('Bagging Classifier hyperparameters:')
print(bagging_classifier_model.get_params())

bagging_classifier_model.fit(X_train_processed, y_train)
bagging_classifier_predictions = bagging_classifier_model.predict(X_test_processed)

print('Bagging Classifier has been fitted successfully.')
print('Bagging Classifier predictions have been generated.')
print(f'Prediction shape: {bagging_classifier_predictions.shape}')
print('Predicted class counts (not evaluation metrics):')
print(pd.Series(bagging_classifier_predictions).value_counts().sort_index().to_string())
```

### Task 3 conclusion

Four fitted models and their test-set class predictions are now stored for later comparison. No performance metric or confusion matrix has been calculated in this task.

# Task 4 - Model Evaluation

This section evaluates the four Task 3 models on the same processed test set. The positive class is `1`, representing employee attrition (`Yes`).

### Evaluation metrics

- **Accuracy:** proportion of all test predictions that are correct.
- **Precision:** among employees predicted to leave, the proportion who actually leave.
- **Recall:** among employees who actually leave, the proportion correctly identified.
- **F1-score:** harmonic mean of precision and recall.
- **ROC-AUC:** ability to rank attrition cases above non-attrition cases across classification thresholds; it is calculated from probability scores, not class labels.
- **Confusion matrix:** counts true negatives, false positives, false negatives, and true positives.

```python
# Import only the evaluation functions required for Task 4.
from sklearn.metrics import (
    accuracy_score,
    confusion_matrix,
    f1_score,
    precision_score,
    recall_score,
    roc_auc_score
)

print('=' * 70)
print('TASK 4 - EVALUATION SETUP')
print('=' * 70)

required_task3_objects = [
    'logistic_regression_model', 'logistic_regression_predictions',
    'decision_tree_model', 'decision_tree_predictions',
    'random_forest_model', 'random_forest_predictions',
    'bagging_classifier_model', 'bagging_classifier_predictions',
    'X_test_processed', 'y_test', 'RANDOM_STATE'
]
missing_task3_objects = [
    object_name for object_name in required_task3_objects
    if object_name not in globals()
]

if missing_task3_objects:
    raise RuntimeError(
        'Run all Task 1, Task 2, and Task 3 cells before Task 4. '
        f'Missing objects: {missing_task3_objects}'
    )

if RANDOM_STATE != 42:
    raise ValueError(f'Expected RANDOM_STATE to be 42, found {RANDOM_STATE}.')

print(f'X_test_processed shape: {X_test_processed.shape}')
print(f'y_test shape: {y_test.shape}')
print('Test-set class counts:')
print(y_test.value_counts().sort_index().to_string())
```

```python
# Generate probability scores from the already fitted Task 3 models.
# The probability for class 1 is required for correct ROC-AUC calculation.
print('=' * 70)
print('TASK 4.1 - PROBABILITY SCORE GENERATION')
print('=' * 70)

def get_positive_class_probabilities(fitted_model, features):
    """Return probabilities for the positive attrition class (1)."""
    if 1 not in fitted_model.classes_:
        raise ValueError('The fitted model does not contain positive class 1.')

    positive_class_index = list(fitted_model.classes_).index(1)
    return fitted_model.predict_proba(features)[:, positive_class_index]

logistic_regression_probability_scores = get_positive_class_probabilities(
    logistic_regression_model, X_test_processed
)
decision_tree_probability_scores = get_positive_class_probabilities(
    decision_tree_model, X_test_processed
)
random_forest_probability_scores = get_positive_class_probabilities(
    random_forest_model, X_test_processed
)
bagging_classifier_probability_scores = get_positive_class_probabilities(
    bagging_classifier_model, X_test_processed
)

probability_score_sets = {
    'Logistic Regression': logistic_regression_probability_scores,
    'Decision Tree': decision_tree_probability_scores,
    'Random Forest': random_forest_probability_scores,
    'Bagging Classifier': bagging_classifier_probability_scores
}

for model_name, probability_scores in probability_score_sets.items():
    if len(probability_scores) != len(y_test):
        raise ValueError(
            f'{model_name} probability scores do not match the test-set length.'
        )
    print(f'{model_name}: probability scores generated with shape {probability_scores.shape}.')
```

```python
# Calculate the required evaluation metrics from Task 3 predictions and Task 4 probabilities.
print('=' * 70)
print('TASK 4.2 - METRICS AND CONFUSION MATRICES')
print('=' * 70)

model_evaluation_inputs = {
    'Logistic Regression': {
        'predictions': logistic_regression_predictions,
        'probabilities': logistic_regression_probability_scores
    },
    'Decision Tree': {
        'predictions': decision_tree_predictions,
        'probabilities': decision_tree_probability_scores
    },
    'Random Forest': {
        'predictions': random_forest_predictions,
        'probabilities': random_forest_probability_scores
    },
    'Bagging Classifier': {
        'predictions': bagging_classifier_predictions,
        'probabilities': bagging_classifier_probability_scores
    }
}

evaluation_results = []
confusion_matrices = {}

for model_name, model_outputs in model_evaluation_inputs.items():
    predictions = model_outputs['predictions']
    probability_scores = model_outputs['probabilities']

    if len(predictions) != len(y_test):
        raise ValueError(f'{model_name} predictions do not match the test-set length.')

    model_confusion_matrix = confusion_matrix(y_test, predictions, labels=[0, 1])
    confusion_matrices[model_name] = model_confusion_matrix

    evaluation_results.append({
        'Model': model_name,
        'Accuracy': accuracy_score(y_test, predictions),
        'Precision': precision_score(y_test, predictions, pos_label=1, zero_division=0),
        'Recall': recall_score(y_test, predictions, pos_label=1, zero_division=0),
        'F1-score': f1_score(y_test, predictions, pos_label=1, zero_division=0),
        'ROC-AUC': roc_auc_score(y_test, probability_scores)
    })

    print(f'{model_name}: required metrics and confusion matrix calculated.')

comparison_table = (
    pd.DataFrame(evaluation_results)
    .set_index('Model')
    .sort_values(by='F1-score', ascending=False)
    .round(4)
)

print('\nModel comparison table:')
print(comparison_table.to_string())

for model_name, model_confusion_matrix in confusion_matrices.items():
    confusion_matrix_table = pd.DataFrame(
        model_confusion_matrix,
        index=['Actual No (0)', 'Actual Yes (1)'],
        columns=['Predicted No (0)', 'Predicted Yes (1)']
    )
    print(f'\nConfusion matrix - {model_name}:')
    print(confusion_matrix_table.to_string())
```

### Discussion of model strengths, weaknesses, and use cases

- **Logistic Regression:** Its main strength is interpretability: the model has a direct coefficient for each processed feature. It may be preferred when HR needs a clear, explainable baseline. Its weakness is that it models a linear relationship in the transformed features and may miss more complex interactions.
- **Decision Tree:** Its rule-based structure is easy to explain and visualise. However, a single tree can be sensitive to the particular training sample and can miss patterns that require multiple rules or interactions. It may be preferred when straightforward decision rules are especially valuable.
- **Random Forest:** Combining many trees can model nonlinear relationships and usually reduces the instability of a single tree. Its weaknesses are reduced interpretability and higher computational cost. It may be preferred when predictive performance is more important than explaining one simple set of rules.
- **Bagging Classifier:** Bagging reduces the variance of individual decision trees by averaging many bootstrap-trained estimators. Its weaknesses are lower interpretability than one tree and the possibility that it captures less feature diversity than a random forest. It may be preferred when a more stable tree-based model is desired while retaining a controlled base-tree design.

The table and confusion matrices above provide the evidence for comparing these trade-offs. No final model recommendation is made here; that decision belongs to Task 5.
